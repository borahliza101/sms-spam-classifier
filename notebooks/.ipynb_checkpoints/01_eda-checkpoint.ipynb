```python
df.rename(columns={
    'v1' : 'label',
    'v2' : 'message'
}, inplace=True)  
```

```python
df["label"].value_counts()
```

```python
df['message_length'] = df['message'].apply(len)
```

```python
df[['label','message_length']].head()
```

```python
df.groupby('label')['message_length'].mean()
```

```python
df.groupby('label')['message_length'].describe()
```

```python
import matplotlib.pyplot as plt

df.hist(column='message_length', by='label', figsize=(10,5))
plt.show()
```
